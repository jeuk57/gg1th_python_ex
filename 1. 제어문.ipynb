### 리스트 컴프리헨션

```python
array = [i * i for i in range(0, 20, 2)]
array

# 예측 결과 : 
```

```python
# array = [i * i for i in range(0, 20, 2)]
# for 문으로 만들기
array = []

for i in range(0, 20, 2):
    array.append(i * i)

array
```

### 리스트 컴프리헨션의 조건문

```python
# 짝수만 출력하기 
even_nums = []
for i in range(1, 20+1):
    # i가 짝수인지 판단, 짝수이면, 리스트에 추가
    if i % 2 == 0:
        even_nums.append(i)
even_nums
```

```python
array = [i for i in range(1, 20+1) if i % 2 == 0]
array
```

### 제어문 종합 문제

```python
decimal = []

for num in range(1, 101):
    # 2는 무조건 넣기
    if num == 2:
        decimal.append(num)
    deno = num - 1

    # 한 번이라도 나눠서 떨어지면 반복문 빠져나오기 
    for i in range(2, deno+1):
        if num % i == 0:
            break
        # 최종 1번만 소수에 넣기
        elif i == deno:
            decimal.append(num)

# 소수 판별 최종 결과 출력
print("1~100까지의 소수 결과\n", decimal)
```

```python
decimal = []

for num in range(1, 101):

    if num == 2:
        decimal.append(num)
    deno = num - 1

    for i in range(2, deno+1):
        if num % i == 0:
            break
        elif i == deno:
            decimal.append(num)

print("1~100까지의 소수 결과\n", decimal)
```
